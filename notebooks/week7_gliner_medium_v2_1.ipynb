```python
from gliner import GLiNER

model = GLiNER.from_pretrained("urchade/gliner_medium-v2.1")

text = """Hi team, just a quick update on the upcoming Community Health Outreach Drive scheduled for next Thursday, November 19th. We will be setting up the main screening tents at the Nakuru Central Community Center starting at 8:00 AM. Dr. Sarah Mwangi, our lead physician, along with volunteers John Omondi and Grace Kariuki, will be overseeing the maternal health checks. Please ensure that all medical supplies are loaded into the transport van by Wednesday evening. The VP of Operations, Michael Ndungu, plans to visit the site around noon to evaluate the workflow. We expect over 200 residents to attend, so please confirm your availability in the chat by Monday.

Field Report for October 12, 2026: Today, our mobile clinic unit traveled to the rural outskirts of Narok. I conducted home visits for five families. At 10:30 AM, I met with a patient named Florence Adhiambo, whose contact number is 0712-345-678, to discuss her ongoing prenatal care and delivery plan. We later held a small town hall gathering at the Kajiado East Primary School at 2:00 PM to distribute mosquito nets and malaria medication. During the session, David Kipkemboi and Lucy Wanjiku reported some mild side effects from the previous batch of supplements. I have logged their specific symptoms and national ID numbers in the secure portal. We will follow up with them during the Weekly Health Check-in next Monday.

Attention all Dandelion Africa program managers. The Q3 Resource Allocation Summit has been moved to Friday, October 30th, taking place in Conference Room B at the Nairobi Regional Office. The summit will begin promptly at 9:15 AM. Attendees must include our Regional Director, Dr. Samuel Limo, the Chief Financial Officer, Alice Mutuku, and all active field supervisors. We will be discussing the budget for the new sanitation facilities in Baringo County and reviewing the latest WhatsApp data exports to see where resources are needed most. Please forward your preliminary expense reports to our logistics coordinator, Peter Otieno, at potieno@example-dandelion.org before the close of business on Wednesday."""
labels = [ "Person", "Job Title", "Date", "Time", "Location", "Event Name", "Phone Number", "ID Number", "Email" ]

entities = model.predict_entities(text, labels)
# for entity in entities:
#    print(entity["text"], "=>", entity["label"])

# look for names to censor
sensitive_labels = ["Person", "Phone Number", "ID Number", "Email"]
sensitive_entities = [e for e in entities if e["label"] in sensitive_labels]
sensitive_entities = sorted(sensitive_entities, key=lambda x: x["start"], reverse=True)

# censor names
redacted_text = text
for entity in sensitive_entities:
  start = entity["start"]
  end = entity["end"]
  redacted_text = redacted_text[:start] + "[REDACTED]" + redacted_text[end:]

print("----- REDACTED LABELS -----")
for e in entities:
  new_text = "[REDACTED]" if e["label"] in sensitive_labels else e["text"]
  print(f"{e["label"]}: {new_text}")

print("\n----- REDACTED TEXT -----")
print(redacted_text)

# convert labels/text into json
import json
import datetime
entry_data = {}
for e in entities:
  label = e["label"]
  text = e["text"]

  # put same labels together
  if label in entry_data:
    if isinstance(entry_data[label], list):
      entry_data[label].append(text)
    else:
      entry_data[label] = [entry_data[label], text]
  else:
    entry_data[label] = text

# build json
database_entry = {
  "entry_id": f"log_{datetime.datetime.now().strftime('%Y%m%d%H%M%S')}",
  "extracted_attributes": entry_data,
  "redacted_message": redacted_text.strip()
}

print("\n----- TEXT TO JSON -----")
print(json.dumps(database_entry, indent=2))
```
